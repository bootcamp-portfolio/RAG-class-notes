### Wikipedia Reader

```python
# !pip install wikipedia llama-index-readers-wikipedia
```

```python
import wikipedia
from llama_index.core import Document
```

```python
wikipedia.set_lang("en")
page = wikipedia.page("Python (programming language)", auto_suggest = False)
documents = [
    Document(
        text = page.content,
        metadata =\
            {
                "title": page.title,
                "url": page.url
            }
    )
]
print(documents)
```
