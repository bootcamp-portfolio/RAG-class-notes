### 1. Document 객체 직접 생성

```python
from llama_index.core import Document

text = """
AI 챗봇 사용 가이드:
- 간단한 질문으로 시작하세요
- 구체적인 예시를 포함하면 더 좋은 답변을 받을 수 있습니다.
- 부적절한 내용은 자동으로 필터링 됩니다.
"""
```

```python
document = Document(
    text = text,
    metadata =\
        {
            "author": "제품개발팀",
            "category": "사용설명서",
            "version": "1.0",
        },
        id = "manual_ai_001"
)

print(document)
```
