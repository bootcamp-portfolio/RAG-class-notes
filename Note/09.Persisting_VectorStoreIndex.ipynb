```python
from llama_index.core import VectorStoreIndex, SimpleDirectoryReader
from llama_index.llms.ollama import Ollama
from llama_index.embeddings.ollama import OllamaEmbedding
```

```python
# LLM과 임베딩 모델 설정
llm = Ollama(
    model = "gemma2:2b",
    temperature = 0,
    request_timeout = 120.0
)

embed_model = OllamaEmbedding(
    model_name = "nomic-embed-text",
)
```

```python
# 데이터 로드
documents = SimpleDirectoryReader("../Data/pdf_sample2/").load_data()
```

```python
# 인덱스 생성 및 데이터 임베딩
index = VectorStoreIndex.from_documents(
    documents = documents,
    embed_model = embed_model,
    show_progress = True
)
```

```python
# 벡터 DB 관련 라마인덱스 패키지
from llama_index.core import StorageContext, load_index_from_storage

# 저장할 디렉토리
persist_dir = "./saved_index"

# 인덱스 저장
index.storage_context.persist(persist_dir = persist_dir)
```

```python
# 저장된 인덱스 로드
storage_context = StorageContext.from_defaults(persist_dir = persist_dir)
```

```python
# 인덱스 로드
loaded_index = load_index_from_storage(
    storage_context = storage_context,
    embed_model = embed_model
)
```

```python
# 쿼리 엔진
loaded_query_engine = loaded_index.as_query_engine(llm = llm)
```

```python
# 쿼리 실행
query = "what is advantage of models in thie documents?"
response = loaded_query_engine.query(query)

# 응답 출력
print(response)
```

```python
# 쿼리 실행
query = "이 논문에서 제안하는 모델의 장점은 무엇인가요? 답변은 한국어로 해주세요"
response = loaded_query_engine.query(query)

# 응답 출력
print(response)
```

### 메모리 부분
